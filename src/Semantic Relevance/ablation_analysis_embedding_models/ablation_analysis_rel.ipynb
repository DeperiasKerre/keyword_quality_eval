```python
import json
import pandas as pd

INPUT_FILE = "semantic_relevance_results.json"
OUTPUT_CSV = "semantic_relevance_model_comparison.csv"


# ---------------------------
# LOAD RESULTS
# ---------------------------
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)


# ---------------------------
# BUILD COMPARISON TABLE
# ---------------------------
rows = []

for model_name, results in data.items():
    df = pd.DataFrame(results)

    rows.append({
        "Embedding Model": model_name.split("/")[-1],

        "Agg Title Mean": round(df["agg_title_similarity"].mean(), 4),
        "Agg Title Median": round(df["agg_title_similarity"].median(), 4),

        "Agg Description Mean": round(df["agg_description_similarity"].mean(), 4),
        "Agg Description Median": round(df["agg_description_similarity"].median(), 4),

        "Avg Title Mean": round(df["avg_title_similarity"].mean(), 4),
        "Avg Title Median": round(df["avg_title_similarity"].median(), 4),

        "Avg Description Mean": round(df["avg_description_similarity"].mean(), 4),
        "Avg Description Median": round(df["avg_description_similarity"].median(), 4)
    })

comparison_df = pd.DataFrame(rows)

comparison_df = comparison_df.sort_values("Embedding Model")


# ---------------------------
# OUTPUT
# ---------------------------
print("\nSemantic Relevance Robustness Comparison\n")
print(comparison_df.to_string(index=False))

comparison_df.to_csv(OUTPUT_CSV, index=False)

print(f"\nCSV saved to: {OUTPUT_CSV}")
```
