```python
import json
import pandas as pd

INPUT_FILE = "coverage_extractor_results.json"

OUTPUT_CSV_MAIN = "coverage_extractor_comparison.csv"
OUTPUT_CSV_THRESHOLDS = "coverage_extractor_threshold_robustness.csv"

THRESHOLDS = [0.6, 0.7, 0.8]

# LOAD RESULTS

with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

# TABLE 1: MAIN COMPARISON (T=0.7)

main_rows = []

for extractor_name, results in data.items():
    df = pd.DataFrame(results)

    title_col = "title_coverage_0.7"
    description_col = "description_coverage_0.7"

    main_rows.append({
        "Concept Extractor": extractor_name,

        "Title Coverage Mean": round(df[title_col].mean(), 4),
        "Title Coverage Median": round(df[title_col].median(), 4),

        "Description Coverage Mean": round(df[description_col].mean(), 4),
        "Description Coverage Median": round(df[description_col].median(), 4)
    })

main_df = pd.DataFrame(main_rows)
main_df = main_df.sort_values("Concept Extractor")

print("\nConceptual Coverage Robustness Across Extractors (T=0.7)\n")
print(main_df.to_string(index=False))

main_df.to_csv(OUTPUT_CSV_MAIN, index=False)

# TABLE 2: THRESHOLD ROBUSTNESS
threshold_rows = []

extractor_frames = {
    extractor_name: pd.DataFrame(results)
    for extractor_name, results in data.items()
}

for t in THRESHOLDS:
    threshold_rows.append({
        "Threshold": t,

        "YAKE Title": round(
            extractor_frames["YAKE"][f"title_coverage_{t}"].mean(),
            4
        ),

        "YAKE Description": round(
            extractor_frames["YAKE"][f"description_coverage_{t}"].mean(),
            4
        ),

        "KeyBERT Title": round(
            extractor_frames["KeyBERT"][f"title_coverage_{t}"].mean(),
            4
        ),

        "KeyBERT Description": round(
            extractor_frames["KeyBERT"][f"description_coverage_{t}"].mean(),
            4
        )
    })

threshold_df = pd.DataFrame(threshold_rows)

print("\nCoverage Threshold Robustness Across Concept Extractors\n")
print(threshold_df.to_string(index=False))

threshold_df.to_csv(OUTPUT_CSV_THRESHOLDS, index=False)

# DONE

print("\nCSV files saved:")
print(f"- {OUTPUT_CSV_MAIN}")
print(f"- {OUTPUT_CSV_THRESHOLDS}")
```
