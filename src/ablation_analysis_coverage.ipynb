```python
import json
import pandas as pd

INPUT_FILE = "coverage_results_rich.json"
OUTPUT_CSV = "coverage_model_comparison.csv"

THRESHOLD = 0.7

title_col = f"title_coverage_{THRESHOLD}"
description_col = f"description_coverage_{THRESHOLD}"


# ---------------------------
# LOAD RESULTS
# ---------------------------
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)


# ---------------------------
# EXTRACT MODEL RESULTS
# ---------------------------
model_summaries = {}

for model_name, results in data.items():
    df = pd.DataFrame(results)

    short_name = model_name.split("/")[-1]

    model_summaries[short_name] = {
        "Title Mean": round(df[title_col].mean(), 4),
        "Title Median": round(df[title_col].median(), 4),
        "Description Mean": round(df[description_col].mean(), 4),
        "Description Median": round(df[description_col].median(), 4)
    }


# ---------------------------
# BUILD SINGLE COMPARISON TABLE
# ---------------------------
comparison_df = pd.DataFrame({
    "Metric": [
        "Mean",
        "Median"
    ],

    "all-MiniLM-L6-v2 Title": [
        model_summaries["all-MiniLM-L6-v2"]["Title Mean"],
        model_summaries["all-MiniLM-L6-v2"]["Title Median"]
    ],

    "all-MiniLM-L6-v2 Description": [
        model_summaries["all-MiniLM-L6-v2"]["Description Mean"],
        model_summaries["all-MiniLM-L6-v2"]["Description Median"]
    ],

    "all-mpnet-base-v2 Title": [
        model_summaries["all-mpnet-base-v2"]["Title Mean"],
        model_summaries["all-mpnet-base-v2"]["Title Median"]
    ],

    "all-mpnet-base-v2 Description": [
        model_summaries["all-mpnet-base-v2"]["Description Mean"],
        model_summaries["all-mpnet-base-v2"]["Description Median"]
    ]
})


# ---------------------------
# OUTPUT
# ---------------------------
print(f"\nConceptual Coverage Robustness Comparison (T={THRESHOLD})\n")
print(comparison_df.to_string(index=False))

comparison_df.to_csv(OUTPUT_CSV, index=False)

print(f"\nCSV saved to: {OUTPUT_CSV}")
```
