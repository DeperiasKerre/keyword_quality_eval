```python
import json
import pandas as pd

INPUT_FILE = "coverage_extractor_results.json"

OUTPUT_CSV = "coverage_extractor_threshold_robustness_full.csv"

THRESHOLDS = [0.6, 0.7, 0.8]

# LOAD RESULTS

with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

# CREATE DATAFRAMES


extractor_frames = {
    extractor_name: pd.DataFrame(results)
    for extractor_name, results in data.items()
}

yake = extractor_frames["YAKE"]
keybert = extractor_frames["KeyBERT"]

# THRESHOLD ROBUSTNESS
# (MEAN + MEDIAN)

rows = []

for t in THRESHOLDS:

    yake_title = yake[f"title_coverage_{t}"]
    yake_desc = yake[f"description_coverage_{t}"]

    keybert_title = keybert[f"title_coverage_{t}"]
    keybert_desc = keybert[f"description_coverage_{t}"]

    # Mean row
    rows.append({
        "Threshold": t,
        "Statistic": "Mean",

        "YAKE Title": round(yake_title.mean(), 4),
        "YAKE Description": round(yake_desc.mean(), 4),

        "KeyBERT Title": round(keybert_title.mean(), 4),
        "KeyBERT Description": round(keybert_desc.mean(), 4)
    })

    # Median row
    rows.append({
        "Threshold": t,
        "Statistic": "Median",

        "YAKE Title": round(yake_title.median(), 4),
        "YAKE Description": round(yake_desc.median(), 4),

        "KeyBERT Title": round(keybert_title.median(), 4),
        "KeyBERT Description": round(keybert_desc.median(), 4)
    })

# OUTPUT

threshold_df = pd.DataFrame(rows)

print("\nSemantic Completeness Threshold Robustness Across Extractors\n")
print(threshold_df.to_string(index=False))

threshold_df.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("\nCSV file saved:")
print(f"- {OUTPUT_CSV}")
```
