```python
import json
import random

# -----------------------------
# Loading the full dataset
# -----------------------------
with open("eu_datasets_cleaned.json", "r", encoding="utf-8") as f:
    datasets = json.load(f)

# -----------------------------
# Safety check
# -----------------------------
if len(datasets) < 60:
    raise ValueError("Dataset must contain at least 60 entries.")

# -----------------------------
# Random sampling
# -----------------------------
random.seed(42)  # for reproducibility

sample_60 = random.sample(datasets, 60)

# -----------------------------
# Split
# -----------------------------
experiment_set = sample_60[:35]
gold_set = sample_60[35:]

# -----------------------------
# Save files
# -----------------------------
with open("experiment_35.json", "w", encoding="utf-8") as f:
    json.dump(experiment_set, f, indent=2, ensure_ascii=False)

with open("gold_25.json", "w", encoding="utf-8") as f:
    json.dump(gold_set, f, indent=2, ensure_ascii=False)

print("Dataset split complete.")
```

```python
#confirming the dataset sizes
print(f"Experiment set: {len(experiment_set)} datasets")
print(f"Gold set: {len(gold_set)} datasets")
```
