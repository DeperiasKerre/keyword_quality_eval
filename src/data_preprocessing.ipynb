```python
#loading json data
import json
with open(r"D:\Post Doc Research\Meta Data\eu_datasets_raw.json", "r",  encoding="utf-8") as file:
    datasets = json.load(file)
```

```python
#Normalizing blank node licenses
import json

# -----------------------------
# Clean license function
# -----------------------------
def clean_license(license_value):
    if not license_value:
        return None

    # Remove blank node licenses
    if license_value.startswith("nodeID://"):
        return None

    return license_value  # keep original URI
# -----------------------------
# Clean licenses
# -----------------------------
for d in datasets:
    d["license"] = clean_license(d.get("license"))

# -----------------------------
# Save cleaned file
# -----------------------------
with open("eu_datasets_cleaned.json", "w", encoding="utf-8") as f:
    json.dump(datasets, f, indent=2, ensure_ascii=False)

print("License cleaning complete (nodeID removed).")
```
