```python
import json
import time
from sentence_transformers import SentenceTransformer, util


INPUT_FILE = "dataset_sample_10k.json"
OUTPUT_FILE = "semantic_relevance_results.json"

#LOAD EMBEDDING MODEL
model = SentenceTransformer("all-MiniLM-L6-v2")

#KEYWORDS NORMALIZATION
def normalize_keywords(keywords):
    if isinstance(keywords, list):
        return [k.lower().strip() for k in keywords if isinstance(k, str)]
    elif isinstance(keywords, str):
        return [keywords.lower().strip()]
    return []

def build_keyword_text(keywords):
    return " ".join(keywords)

def get_content_text(item):
    return f"{item.get('title', '')} {item.get('description', '')}"

# -------- MAIN --------
if __name__ == "__main__":

    start_time = time.time()

    print("🔄 Loading dataset...\n")
    with open(INPUT_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    print(f"Loaded {len(data)} datasets\n")

    # -------- PREPARE TEXT --------
    content_texts = []
    keyword_texts = []
    keyword_lists = []

    for d in data:
        keywords = normalize_keywords(d.get("keywords", []))
        keyword_lists.append(keywords)

        keyword_texts.append(build_keyword_text(keywords))
        content_texts.append(get_content_text(d))

    # -------- EMBEDDINGS --------
    print("🔄 Computing embeddings...")

    embed_start = time.time()

    content_embeddings = model.encode(content_texts, convert_to_tensor=True, batch_size=64)
    keyword_embeddings = model.encode(keyword_texts, convert_to_tensor=True, batch_size=64)

    embed_end = time.time()

    print(f"✅ Embeddings computed in {embed_end - embed_start:.2f}s\n")

    # -------- SIMILARITY --------
    print("🔄 Computing semantic relevance...\n")

    results = []

    for i in range(len(data)):

        content_emb = content_embeddings[i]
        keyword_emb = keyword_embeddings[i]

        # ---- Aggregated similarity ----
        agg_sim = float(util.cos_sim(keyword_emb, content_emb)[0][0])

        # ---- Per-keyword similarity ----
        keywords = keyword_lists[i]

        if len(keywords) > 0:
            keyword_embs = model.encode(keywords, convert_to_tensor=True, batch_size=32)
            sims = util.cos_sim(keyword_embs, content_emb).cpu().numpy()
            avg_sim = float(sims.mean())
        else:
            avg_sim = 0.0

        results.append({
            "dataset_id": data[i].get("dataset_id"),
            "agg_similarity": agg_sim,
            "avg_keyword_similarity": avg_sim,
            "num_keywords": len(keywords)
        })

        if (i + 1) % 500 == 0:
            print(f"Processed {i+1}/{len(data)} datasets")

    # -------- SAVE RESULTS --------
    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        json.dump(results, f, indent=2)

    print(f"\n💾 Results saved to {OUTPUT_FILE}")

    end_time = time.time()
    print(f"\n⏱ Total time: {end_time - start_time:.2f}s")
```
